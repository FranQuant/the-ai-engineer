
# Finance MiniGPT: What Can a Tiny Transformer Learn From FOMC Statements?

This capstone builds a decoder-only Transformer from first principles and trains it on official FOMC policy statements from 2015–2025. The experiment asks a deliberately narrow question: **can a small character-level model learn the statistical form of central-bank language, and where does that learning stop?**

The notebook follows the experiment in the order a reader needs: FOMC corpus → autoregressive task → causal attention → Finance MiniGPT → training protocol → canonical results → generated text → honest evaluation.

## Headline result

The model learned measurable character-level and stylistic regularities, and its chronological-validation loss improved under a controlled training protocol. Its samples nevertheless remained semantically unreliable. The result is therefore evidence of next-character modeling—not evidence of policy understanding, economic reasoning, or investable forecasting skill.

## 2. FOMC corpus and chronological design

An individual FOMC statement is the fundamental document unit because it is a dated policy communication with its own decision context and provenance. Treating statements as documents preserves chronology and prevents a training example from splicing the end of one meeting’s statement to the beginning of another.

The committed corpus contains 90 normalized official statements. The loader verifies the corpus bytes and hash, manifest identity, extraction and normalization versions, document headers and bodies, chronological ordering, split labels, and per-document hashes before returning any modeling text.

The whole-document split is fixed:

- **Training:** 82 statements from 2015–2024.
- **Validation:** 8 statements from 2025.

Context windows may overlap within a statement, but they never cross statement boundaries. The 2025 loss was inspected repeatedly for checkpoint selection, so 2025 is validation data—not a pristine test set or untouched out-of-sample test. Overlapping windows are also not independent observations.


```python
EXPECTED = {
    "corpus_id": "fomc-statements-2015-2025-v2",
    "sha256": "c632b6a2e7bcfc0360e9fe18113a2ec1c9edce2d42f83b4ff96f5ce4d74ea125",
    "bytes": 265669, "characters": 265669, "documents": 90,
    "train": 82, "validation": 8, "extraction_method": "federalreserve_article_paragraphs_v2",
    "normalization_version": "fomc-finance-preserving-v1",
}
START, END = "<|fomc_statement|>", "<|end_fomc_statement|>"

def resolve_corpus_dir() -> Path:
    starts = [Path.cwd()]
    if os.environ.get("WEEK03_REPO_ROOT"): starts.append(Path(os.environ["WEEK03_REPO_ROOT"]).resolve())
    if "__file__" in globals(): starts.append(Path(__file__).resolve().parent)
    for start in starts:
        for parent in [start, *start.parents]:
            for candidate in (parent, parent / "capstones" / "week03_transformers"):
                if (candidate / "fomc_statements_2015_2025.txt").is_file() and (candidate / "fomc_statements_2015_2025_manifest.json").is_file():
                    return candidate
    raise FileNotFoundError("Frozen FOMC corpus files not found from cwd/notebook parents.")

def sha256_text(text: str) -> str:
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

def load_verified_corpus(corpus_dir: Path):
    corpus_path = corpus_dir / "fomc_statements_2015_2025.txt"
    manifest_path = corpus_dir / "fomc_statements_2015_2025_manifest.json"
    raw_bytes = corpus_path.read_bytes()
    text = raw_bytes.decode("utf-8")
    manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
    checks = {
        "sha256": hashlib.sha256(raw_bytes).hexdigest() == EXPECTED["sha256"] == manifest["corpus_sha256"],
        "bytes": len(raw_bytes) == EXPECTED["bytes"] == manifest["corpus_utf8_bytes"],
        "characters": len(text) == EXPECTED["characters"] == manifest["corpus_characters"],
        "corpus_id": manifest["corpus_id"] == EXPECTED["corpus_id"],
        "extraction_method": manifest["extraction_method"] == EXPECTED["extraction_method"],
        "normalization_version": manifest["normalization_version"] == EXPECTED["normalization_version"],
        "document_count": manifest["document_count"] == EXPECTED["documents"] == len(manifest["documents"]),
        "split_counts": manifest["train_document_count"] == EXPECTED["train"] and manifest["validation_document_count"] == EXPECTED["validation"],
        "zero_failures": len(manifest["failures"]) == 0,
        "marker_counts": text.count(START) == text.count(END) == EXPECTED["documents"],
    }
    failed = [name for name, ok in checks.items() if not ok]
    if failed: raise ValueError(f"Corpus integrity mismatch: {failed}")

    pattern = re.compile(
        rf"{re.escape(START)}\n"
        r"date: (?P<date>\d{4}-\d{2}-\d{2})\n"
        r"meeting_type: (?P<meeting_type>scheduled|unscheduled)\n"
        r"document_id: (?P<document_id>[^\n]+)\n\n"
        rf"(?P<body>.*?)(?=\n{re.escape(END)})\n{re.escape(END)}",
        re.DOTALL,
    )
    matches = list(pattern.finditer(text))
    reconstructed = "\n\n".join(m.group(0) for m in matches) + "\n"
    if len(matches) != EXPECTED["documents"] or reconstructed != text:
        raise ValueError("Structural marker parse was not lossless.")
    records = []
    for match, meta in zip(matches, manifest["documents"]):
        record = {"document_id": match["document_id"], "statement_date": match["date"], "meeting_type": match["meeting_type"], "split": meta["split"], "body": match["body"]}
        expected_fields = (meta["document_id"], meta["statement_date"], meta["meeting_type"])
        if (record["document_id"], record["statement_date"], record["meeting_type"]) != expected_fields:
            raise ValueError(f"Document order/header mismatch at {meta['document_id']}")
        if len(record["body"]) != meta["normalized_characters"] or sha256_text(record["body"]) != meta["normalized_sha256"]:
            raise ValueError(f"Body integrity mismatch for {record['document_id']}")
        expected_split = "train" if record["statement_date"] <= "2024-12-31" else "validation"
        if record["split"] != expected_split or (record["split"] == "validation") != record["statement_date"].startswith("2025-"):
            raise ValueError(f"Split/date mismatch for {record['document_id']}")
        records.append(record)
    if [r["statement_date"] for r in records] != sorted(r["statement_date"] for r in records):
        raise ValueError("Documents are not in chronological order.")
    return records, manifest, checks

CORPUS_DIR = resolve_corpus_dir()
documents, manifest, integrity_checks = load_verified_corpus(CORPUS_DIR)
print("Corpus integrity PASS:", all(integrity_checks.values()), "|", CORPUS_DIR)

# ---- preserved source boundary ----

train_docs = [d for d in documents if d["split"] == "train"]
val_docs = [d for d in documents if d["split"] == "validation"]
assert {d["document_id"] for d in train_docs}.isdisjoint(d["document_id"] for d in val_docs)
assert len(train_docs) == 82 and len(val_docs) == 8
assert sum(len(d["body"]) for d in train_docs) == 237840
assert sum(len(d["body"]) for d in val_docs) == 17474
assert Counter(d["meeting_type"] for d in documents) == {"scheduled": 87, "unscheduled": 3}

def available_windows(records, block_size): return sum(max(0, len(d["body"]) - block_size) for d in records)
summary_rows = []
for name, records in (("train", train_docs), ("validation", val_docs)):
    summary_rows.append({"split": name, "documents": len(records), "date_range": f"{records[0]['statement_date']} to {records[-1]['statement_date']}", "body_characters": sum(len(d['body']) for d in records), "windows@128": available_windows(records, 128)})

print("Corpus identity")
print(f"  ID: {manifest['corpus_id']}")
print(f"  SHA-256: {manifest['corpus_sha256']}")
print(f"  Documents: {len(documents)} (87 scheduled, 3 unscheduled)")
print("\nChronological split (whole statements; context length 128)")
print(f"{'split':<12}{'statements':>12}{'body chars':>14}{'windows':>12}  date range")
for row in summary_rows:
    print(f"{row['split']:<12}{row['documents']:>12,}{row['body_characters']:>14,}{row['windows@128']:>12,}  {row['date_range']}")

clean_excerpt = " ".join(train_docs[0]["body"].split())[:320]
print("\nCleaned modeling-text excerpt (metadata and serialized markers excluded)")
print(clean_excerpt + "…")
```

## 3. Character-level autoregressive prediction

For a character sequence $x_{1:T}$, the language model factorizes its probability as

$$
p_\theta(x_{1:T}) = \prod_{t=1}^{T} p_\theta\!\left(x_t \mid x_{<t}\right).
$$

The code realizes this equation by shifting the same text by one position: each input prefix asks the model to predict the next character in its target. A character vocabulary keeps tokenization transparent, avoids unknown words, and makes every prediction inspectable. Its cost is longer sequences and no explicit representation of words or monetary-policy concepts; the model must learn such structure indirectly from characters.

The vocabulary is constructed only from training statement bodies. Dates, document IDs, split labels, and serialization markers are excluded. Validation text and the five fixed evaluation prompts must be covered by the training vocabulary.


```python
EVAL_PROMPTS = [
    "The Committee decided to", "Inflation has", "The labor market",
    "The target range for the federal funds rate",
    "In assessing the appropriate stance of monetary policy",
]
training_bodies = tuple(d["body"] for d in train_docs)
training_body_chars = sorted(set().union(*(set(body) for body in training_bodies)))
chars = training_body_chars.copy()
stoi = {ch: i for i, ch in enumerate(chars)}; itos = {i: ch for ch, i in stoi.items()}
VOCAB_SIZE = len(chars)

def encode(text: str) -> list[int]:
    unknown = sorted(set(text) - set(stoi))
    if unknown: raise ValueError(f"Unknown characters: {unknown!r}")
    return [stoi[ch] for ch in text]

def decode(ids: Sequence[int]) -> str:
    try: return "".join(itos[int(i)] for i in ids)
    except KeyError as exc: raise ValueError(f"Token ID outside vocabulary: {exc.args[0]}") from exc

serialized_headers = tuple(
    line
    for d in documents
    for line in (
        START, f'date: {d["statement_date"]}', f'meeting_type: {d["meeting_type"]}',
        f'document_id: {d["document_id"]}', END,
    )
)
model_documents = training_bodies
tokenizer_inputs = training_bodies
marker_exclusion = (
    chars == sorted(set().union(*(set(body) for body in training_bodies)))
    and model_documents == tuple(d["body"] for d in train_docs)
    and tokenizer_inputs == tuple(d["body"] for d in train_docs)
    and all(START not in body and END not in body for body in model_documents)
    and all(header not in body.splitlines() for body in model_documents for header in serialized_headers)
    and all(not body.startswith((START, "date: ", "meeting_type: ", "document_id: ")) for body in model_documents)
)
record_check("body-only vocabulary exactness", chars == training_body_chars)
record_check("structural marker/header exclusion", marker_exclusion)
record_check("training vocabulary size", VOCAB_SIZE == 73)
record_check("tokenizer train round-trip", all(decode(encode(d["body"])) == d["body"] for d in train_docs))
record_check("validation character coverage", not (set().union(*(set(d["body"]) for d in val_docs)) - set(chars)))
record_check("prompt character coverage", all(decode(encode(prompt)) == prompt for prompt in EVAL_PROMPTS))
VOCABULARY_DATA = {"chars": chars, "stoi": stoi}

tokenizer_example = "The Committee"
tokenizer_ids = encode(tokenizer_example)
shift_text = next(body for body in training_bodies if "The Committee" in body)
shift_start = shift_text.index("The Committee")
shift_segment = shift_text[shift_start:shift_start + 9]
input_text, target_text = shift_segment[:-1], shift_segment[1:]
x_small = torch.tensor([encode(input_text)], dtype=torch.long)
y_small = torch.tensor([encode(target_text)], dtype=torch.long)

print(f"Vocabulary size: {VOCAB_SIZE}")
print("Vocabulary:", repr("".join(chars)))
print("\nEncode/decode example")
print("  text:", repr(tokenizer_example))
print("  ids: ", tokenizer_ids)
print("  decoded:", repr(decode(tokenizer_ids)))
print("\nShifted next-character example")
print("  input: ", repr(input_text))
print("  target:", repr(target_text))
print("  x shape:", tuple(x_small.shape), "y shape:", tuple(y_small.shape))
```

## 4. Scaled causal attention

### Motivation, equations, and shapes

For an input representation $X$, learned projections will produce queries, keys, and values:

$$Q=XW_Q, \qquad K=XW_K, \qquad V=XW_V.$$

One attention head then computes

$$
S = \frac{QK^\top}{\sqrt{d_h}} + M, \qquad
A = \operatorname{softmax}(S), \qquad
Y = AV.
$$

These equations directly match the implementation below. $M$ is the causal mask represented operationally by replacing forbidden scores with $-\infty$ before softmax. If query and key components are roughly independent with unit variance, their dot product has variance proportional to $d_h$. Dividing by $\sqrt{d_h}$ stabilizes score magnitudes before softmax and helps avoid needlessly saturated attention weights.

| Tensor | Shape | Role |
|---|---|---|
| $Q$ | `[B,H,T,d_h]` | What each position seeks |
| $K$ | `[B,H,T,d_h]` | What each position offers for matching |
| $V$ | `[B,H,T,d_h]` | Content to mix |
| $S$ | `[B,H,T,T]` | Scaled, causally masked compatibility scores |
| $A$ | `[B,H,T,T]` | Row-normalized attention weights |
| $Y$ | `[B,H,T,d_h]` | Attention-weighted value mixture |

The numerical example uses the existing two-token $Q$, $K$, and $V$ tensors so every intermediate value remains hand-checkable. The following section will learn these tensors from $X$.

### Causal masking

When predicting at position $i$, keys at positions $j>i$ are unavailable. For two tokens, the Boolean allowed-key mask is

$$
\begin{bmatrix}
1 & 0 \\
1 & 1
\end{bmatrix}.
$$

Thus the first query can attend only to the first key, while the second can attend to both. Masking occurs before softmax, so forbidden positions receive exactly zero probability.

### Interpretation and bridge

Scaling preserves the relative dot-product evidence while controlling its magnitude. Causal masking changes only what information is available: the first row becomes $[1,0]$, so token 1 cannot use token 2, while the second row retains access to both keys. Multiplying the masked weights by $V$ produces the verified mixture $Y$.

> **The numerical example treated Q, K, and V as known. The next section makes them learned projections of the input and computes several attention heads in parallel.**


```python
def scaled_dot_product_attention(q, k, v, mask=None, dropout_p=0.0, training=False, return_weights=False):
    if q.ndim != 4 or k.ndim != 4 or v.ndim != 4: raise ValueError("Q, K, V must each have shape [B,H,T,d_head].")
    if q.shape != k.shape or q.shape != v.shape: raise ValueError(f"Q, K, V shapes must match; got {q.shape}, {k.shape}, {v.shape}.")
    if not q.is_floating_point() or q.dtype != k.dtype or q.dtype != v.dtype: raise TypeError("Q, K, V must share a floating dtype.")
    scores = q @ k.transpose(-2, -1) / math.sqrt(q.size(-1))  # [B,H,T,T]
    if mask is not None:
        if mask.dtype != torch.bool: raise TypeError("Attention mask must be boolean (True=allowed).")
        try: allowed = torch.broadcast_to(mask.to(scores.device), scores.shape)
        except RuntimeError as exc: raise ValueError(f"Mask shape {tuple(mask.shape)} is not broadcastable to scores {tuple(scores.shape)}.") from exc
        if not allowed.any(dim=-1).all(): raise ValueError("Every attention row must allow at least one key.")
        scores = scores.masked_fill(~allowed, float("-inf"))
    weights = F.softmax(scores, dim=-1)
    mixed_weights = F.dropout(weights, p=dropout_p, training=training)
    output = mixed_weights @ v  # [B,H,T,d_head]
    return (output, weights) if return_weights else output

# ---- preserved source boundary ----

q_small = torch.tensor([[[[1., 0.], [0., 1.]]]])
k_small = torch.tensor([[[[1., 0.], [0., 1.]]]])
v_small = torch.tensor([[[[1., 2.], [3., 4.]]]])

raw_scores_small = q_small @ k_small.transpose(-2, -1)
scores_small = raw_scores_small / math.sqrt(2)
_, unmasked = scaled_dot_product_attention(q_small, k_small, v_small, return_weights=True)
causal2 = torch.tril(torch.ones(2, 2, dtype=torch.bool))[None, None]
masked_scores_small = scores_small.masked_fill(~causal2, float("-inf"))
out_masked, masked = scaled_dot_product_attention(q_small, k_small, v_small, mask=causal2, return_weights=True)

np_scores = np.array([[1., 0.], [0., 1.]]) / np.sqrt(2.0)
np_shift = np_scores - np_scores.max(axis=-1, keepdims=True)
np_unmasked = np.exp(np_shift) / np.exp(np_shift).sum(axis=-1, keepdims=True)
expected_unmasked = np.array([[0.66976155, 0.33023845], [0.33023845, 0.66976155]])
expected_masked = np.array([[1.0, 0.0], [0.33023845, 0.66976155]])
expected_output = expected_masked @ np.array([[1., 2.], [3., 4.]])

record_check("numerical attention shapes", scores_small.shape == (1,1,2,2) and out_masked.shape == (1,1,2,2))
record_check("independent numerical attention reference", np.allclose(np_unmasked, expected_unmasked, atol=1e-7) and np.allclose(unmasked.numpy()[0,0], expected_unmasked, atol=1e-7) and np.allclose(out_masked.numpy()[0,0], expected_output, atol=1e-7))
record_check("attention rows normalize", torch.allclose(masked.sum(-1), torch.ones_like(masked.sum(-1))))
record_check("future attention is zero", np.allclose(masked.numpy()[0,0], expected_masked, atol=1e-7) and torch.count_nonzero(masked[..., 0, 1]) == 0)
assert not torch.allclose(unmasked[..., 0, :], masked[..., 0, :])
assert torch.allclose(masked[..., 0, :], torch.tensor([[[1.0, 0.0]]]))
assert np.allclose(out_masked.numpy()[0,0], expected_output, atol=1e-7)

def print_matrix(label, tensor):
    print(f"{label}\n{tensor.detach().cpu().numpy()}\n")

print_matrix("Q", q_small[0, 0])
print_matrix("K", k_small[0, 0])
print_matrix("V", v_small[0, 0])
print_matrix("Raw dot products QK^T", raw_scores_small[0, 0])
print_matrix("Scaled scores QK^T / sqrt(d_h)", scores_small[0, 0])
print_matrix("Unmasked attention weights", unmasked[0, 0])
print_matrix("Causal mask (1=allowed, 0=forbidden)", causal2[0, 0].to(torch.int64))
print("Masked scores (-inf denotes a forbidden future key)\n", masked_scores_small[0, 0].numpy(), "\n")
print_matrix("Masked attention weights", masked[0, 0])
print_matrix("Mixed output Y = AV", out_masked[0, 0])
print("Numerical attention evidence PASS: shapes, normalization, masking, first-row change, and weighted mixture")

import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(6.4, 2.6), constrained_layout=True)
for axis, weights, title in zip(axes, (unmasked[0, 0], masked[0, 0]), ("Unmasked weights", "Causal weights")):
    image = axis.imshow(weights.detach().cpu().numpy(), vmin=0, vmax=1, cmap="Blues")
    for row in range(2):
        for column in range(2):
            axis.text(column, row, f"{weights[row, column].item():.3f}", ha="center", va="center", color="black")
    axis.set(title=title, xlabel="key position", ylabel="query position", xticks=[0, 1], yticks=[0, 1])
fig.colorbar(image, ax=axes, shrink=0.82, label="attention weight")
display(fig, metadata={"image/png": {"alt": "Heatmaps comparing unmasked scaled dot-product attention weights with causal masked attention weights; the causal heatmap assigns zero probability to future key positions."}})
plt.close(fig)
```

## 5. Learned multi-head causal self-attention

The model must decide what information is relevant at every layer and position, so it learns the projection matrices that turn the current representation $X$ into queries, keys, and values:

$$Q=XW_Q, \qquad K=XW_K, \qquad V=XW_V.$$

For batch size $B$, sequence length $T$, model width $C=d_{\text{model}}$, head count $H$, and head width $d_h=C/H$, head $h$ computes

$$
\operatorname{head}_h = \operatorname{softmax}\!\left(\frac{Q_hK_h^\top}{\sqrt{d_h}} + M\right)V_h,
$$

and the module recombines the heads through

$$
\operatorname{MHA}(X)=\operatorname{Concat}(\operatorname{head}_1,\ldots,\operatorname{head}_H)W_O.
$$

The canonical implementation below uses one learned linear layer for the concatenated $Q/K/V$ projections, then reshapes it into heads. This is the same module used by every trained Transformer block later in the notebook.

| Canonical route | Shape |
|---|---|
| Input `x` | `[B,T,C]` |
| Joint `qkv(x)` projection | `[B,T,3C]` |
| `view` then `permute` | `[3,B,H,T,d_h]` |
| Each of `q`, `k`, `v` | `[B,H,T,d_h]` |
| Scores and returned weights | `[B,H,T,T]` |
| Attended values by head | `[B,H,T,d_h]` |
| Transpose and concatenate | `[B,T,C]` |
| Output projection | `[B,T,C]` |

The stored mask has shape `[1,1,T_max,T_max]`, so it broadcasts across both batch and head dimensions. Every head remains causal: query position $i$ receives zero probability for keys $j>i$. Attention dropout acts on the weights; residual dropout acts after $W_O$. In evaluation mode both are inactive. Passing `return_attention=True` preserves the weights needed by later inspection without defining a second attention path.


```python
class MultiHeadCausalSelfAttention(nn.Module):
    def __init__(self, d_model, num_heads, dropout, max_seq_len):
        super().__init__()
        if d_model % num_heads: raise ValueError("d_model must be divisible by num_heads.")
        self.num_heads, self.head_dim = num_heads, d_model // num_heads
        self.qkv = nn.Linear(d_model, 3*d_model)
        self.out_proj = nn.Linear(d_model, d_model)
        self.attn_dropout, self.resid_dropout = dropout, nn.Dropout(dropout)
        self.register_buffer("causal_mask", torch.tril(torch.ones(max_seq_len, max_seq_len, dtype=torch.bool))[None,None])
    def forward(self, x, return_attention=False):
        if x.ndim != 3: raise ValueError("Attention input must have shape [B,T,D].")
        B,T,D = x.shape
        if T > self.causal_mask.size(-1): raise ValueError(f"Context length {T} exceeds configured maximum {self.causal_mask.size(-1)}.")
        qkv = self.qkv(x).view(B,T,3,self.num_heads,self.head_dim).permute(2,0,3,1,4)
        q,k,v = qkv.unbind(0)  # each [B,H,T,d_head]
        y, weights = scaled_dot_product_attention(q,k,v,self.causal_mask[:,:,:T,:T],self.attn_dropout,self.training,True)
        y = y.transpose(1,2).contiguous().view(B,T,D)
        y = self.resid_dropout(self.out_proj(y))
        return (y, weights) if return_attention else y

# ---- preserved source boundary ----

seed_everything(0)
mha = MultiHeadCausalSelfAttention(32, 4, 0.0, 16).to(DEVICE)
mha_input = torch.randn(2, 7, 32, device=DEVICE, requires_grad=True)
mha.eval()
mha_output, mha_weights = mha(mha_input, return_attention=True)

record_check("multi-head output shape", mha_output.shape == mha_input.shape)
assert mha_weights.shape == (2, 4, 7, 7)
assert torch.allclose(mha_weights.sum(dim=-1), torch.ones_like(mha_weights.sum(dim=-1)), atol=1e-6)
future = torch.triu(torch.ones(7, 7, dtype=torch.bool, device=DEVICE), diagonal=1)
assert torch.count_nonzero(mha_weights.masked_select(future[None, None])) == 0
assert mha.causal_mask.shape == (1, 1, 16, 16)

mha.zero_grad(set_to_none=True)
mha_output.square().mean().backward()
projection_grads = [parameter.grad for parameter in mha.parameters()]
assert projection_grads and all(gradient is not None and torch.isfinite(gradient).all() for gradient in projection_grads)
with torch.no_grad():
    mha_eval_a = mha(mha_input.detach())
    mha_eval_b = mha(mha_input.detach())
assert torch.equal(mha_eval_a, mha_eval_b)

divisibility_guarded = False
try:
    MultiHeadCausalSelfAttention(30, 4, 0.0, 16)
except ValueError as exc:
    divisibility_guarded = "divisible" in str(exc)
assert divisibility_guarded
print({"input": tuple(mha_input.shape), "weights": tuple(mha_weights.shape), "output": tuple(mha_output.shape), "head_dim": mha.head_dim})
print("Learned multi-head attention evidence PASS: normalized causal weights, projection gradients, eval determinism, and guards")
```

## 6. From positional information to Finance MiniGPT

### 6.1 Sinusoidal positional information

Causal self-attention constrains which positions can interact, but without a position signal it does not identify an absolute token index. Finance MiniGPT therefore adds a deterministic sinusoidal vector to each token embedding. For position $pos$ and channel pair $2i,2i+1$,

$$
\operatorname{PE}(pos,2i)=\sin\!\left(\frac{pos}{10000^{2i/d_{\text{model}}}}\right),
$$

$$
\operatorname{PE}(pos,2i+1)=\cos\!\left(\frac{pos}{10000^{2i/d_{\text{model}}}}\right).
$$

The implementation constructs the equivalent frequency vector with `exp(arange(0,d_model,2) * (-log(10000)/d_model))`, fills alternating sine and cosine channels, and stores `[1,T_max,d_model]` as a buffer. A buffer is non-trainable state: it appears in the state dict and moves with the module device, but the optimizer does not update it. The forward method slices the configured prefix and adds it to token embeddings of shape `[B,T,d_model]`.

The plotted channels oscillate at different frequencies: nearby dimensions provide complementary sine/cosine phases, while frequency changes across channel pairs. The plot is evidence for the exact fixed buffer used by the model, not a learned representation or an attention explanation.

### 6.2 Feed-forward network and Pre-LN Transformer block

Attention mixes information **across sequence positions**. The feed-forward network then transforms each position independently with the same two learned linear maps:

$$
\operatorname{FFN}(x)=W_2\operatorname{GELU}(W_1x+b_1)+b_2.
$$

The canonical module expands from $d_{\text{model}}$ to `d_ff`, applies GELU and dropout, projects back to $d_{\text{model}}`, and applies residual dropout. Matching input/output width is required for the two residual additions. The block uses Pre-LayerNorm ordering:

$$x' = x + \operatorname{Attention}(\operatorname{LN}_1(x)),$$

$$y = x' + \operatorname{FFN}(\operatorname{LN}_2(x')).$$

This ordering gives gradients both residual and transformed paths; it is an architectural choice, not a guarantee of optimization success. The two statements in `forward` below implement these equations in exactly this order.


```python
class SinusoidalPositionalEncoding(nn.Module):
    def __init__(self, d_model, max_seq_len):
        super().__init__()
        if d_model <= 0 or max_seq_len <= 0: raise ValueError("d_model and max_seq_len must be positive.")
        position = torch.arange(max_seq_len, dtype=torch.float32).unsqueeze(1)
        div = torch.exp(torch.arange(0,d_model,2,dtype=torch.float32) * (-math.log(10000.0)/d_model))
        pe = torch.zeros(max_seq_len,d_model); pe[:,0::2] = torch.sin(position*div)
        pe[:,1::2] = torch.cos(position*div[:pe[:,1::2].shape[1]])
        self.register_buffer("pe", pe.unsqueeze(0))
    def forward(self, x):
        if x.ndim != 3: raise ValueError("Positional input must have shape [B,T,D].")
        if x.size(1) > self.pe.size(1): raise ValueError(f"Context length {x.size(1)} exceeds positional maximum {self.pe.size(1)}.")
        if x.size(2) != self.pe.size(2): raise ValueError("Embedding dimension does not match positional encoding.")
        return x + self.pe[:,:x.size(1)].to(dtype=x.dtype)

# ---- preserved source boundary ----

position_demo = SinusoidalPositionalEncoding(32, 16).to(DEVICE)
position_input = torch.zeros(2, 16, 32, device=DEVICE)
position_a = position_demo(position_input)
position_b = position_demo(position_input)
assert position_demo.pe.shape == (1, 16, 32)
assert position_a.shape == position_input.shape
assert not torch.equal(position_a[:, 0], position_a[:, 1])
assert torch.equal(position_a, position_b)
assert position_demo.pe.device == position_input.device
assert not any(name == "pe" for name, _ in position_demo.named_parameters())
assert "pe" in position_demo.state_dict()

import matplotlib.pyplot as plt
selected_dimensions = (0, 1, 2, 3, 6, 7)
fig, axis = plt.subplots(figsize=(7.0, 3.1))
positions = np.arange(position_demo.pe.size(1))
for dimension in selected_dimensions:
    axis.plot(positions, position_demo.pe[0, :, dimension].detach().cpu().numpy(), marker=".", label=f"dim {dimension}")
axis.set(xlabel="position", ylabel="encoding value", title="Sinusoidal positions — selected dimensions")
axis.grid(alpha=0.25); axis.legend(ncol=3, fontsize=8); fig.tight_layout()
display(fig, metadata={"image/png": {"alt": "Line plot of selected sinusoidal positional-encoding dimensions across sequence positions, illustrating deterministic position-dependent variation at different frequencies."}})
plt.close(fig)
print({"buffer": tuple(position_demo.pe.shape), "combined_embeddings": tuple(position_a.shape), "device": str(position_demo.pe.device)})
print("Positional evidence PASS: distinct positions, deterministic calls, device movement, and configured context support")
```


## 7. Build Finance MiniGPT

The complete decoder-only model combines token embeddings, deterministic sinusoidal position information, pre-layer-normalized Transformer blocks, causal self-attention, a final normalization layer, and a tied language-model head. The following preserved implementation and checks establish the architecture, tensor shapes, causal behavior, initialization, weight tying, finite loss, and canonical parameter count.


```python
class TransformerBlock(nn.Module):
    def __init__(self, d_model, num_heads, d_ff, dropout, max_seq_len):
        super().__init__()
        self.ln1, self.ln2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.attn = MultiHeadCausalSelfAttention(d_model,num_heads,dropout,max_seq_len)
        self.ffn = nn.Sequential(nn.Linear(d_model,d_ff), nn.GELU(), nn.Dropout(dropout), nn.Linear(d_ff,d_model), nn.Dropout(dropout))
    def forward(self, x, return_attention=False):
        attn_result = self.attn(self.ln1(x), return_attention=return_attention)
        if return_attention: attn_out, weights = attn_result
        else: attn_out = attn_result
        x = x + attn_out; x = x + self.ffn(self.ln2(x))
        return (x, weights) if return_attention else x

# ---- preserved source boundary ----

seed_everything(0)
block = TransformerBlock(32, 4, 64, 0.0, 16).to(DEVICE)
block.eval()
block_input = torch.randn(2, 7, 32, device=DEVICE, requires_grad=True)
block_output, block_weights = block(block_input, return_attention=True)
record_check("block shape invariance", block_output.shape == block_input.shape)
assert torch.count_nonzero(block_weights.masked_select(future[None, None])) == 0

block.zero_grad(set_to_none=True)
block_output.square().mean().backward()
attention_parameters = [p for name, p in block.named_parameters() if ".attn." in f".{name}"]
ffn_parameters = [p for name, p in block.named_parameters() if ".ffn." in f".{name}"]
assert attention_parameters and ffn_parameters
assert all(p.grad is not None and torch.isfinite(p.grad).all() for p in attention_parameters + ffn_parameters)

causal_a = block_input.detach().clone(); causal_b = causal_a.clone(); causal_b[:, -1] += 1.0
with torch.no_grad():
    causal_out_a = block(causal_a)
    causal_out_b = block(causal_b)
assert torch.allclose(causal_out_a[:, :-1], causal_out_b[:, :-1], atol=1e-7, rtol=0)

dropout_block = TransformerBlock(32, 4, 64, 0.25, 16).to(DEVICE).eval()
with torch.no_grad():
    deterministic_a = dropout_block(causal_a)
    deterministic_b = dropout_block(causal_a)
assert torch.equal(deterministic_a, deterministic_b)
print({"block_input": tuple(block_input.shape), "block_output": tuple(block_output.shape), "attention_weights": tuple(block_weights.shape)})
print("Pre-LN block evidence PASS: shape-preserving residuals, causal behavior, attention/FFN gradients, and eval determinism")

# ---- preserved source boundary ----

class FinanceMiniGPT(nn.Module):
    def __init__(self, vocab_size, architecture):
        super().__init__(); self.vocab_size = vocab_size; self.architecture = dict(architecture)
        D, L, H, FF, BS, P = (architecture[k] for k in ("d_model","num_layers","num_heads","d_ff","block_size","dropout"))
        self.token_embedding = nn.Embedding(vocab_size,D)
        self.position = SinusoidalPositionalEncoding(D,BS); self.dropout = nn.Dropout(P)
        self.blocks = nn.ModuleList([TransformerBlock(D,H,FF,P,BS) for _ in range(L)])
        self.ln_f = nn.LayerNorm(D); self.lm_head = nn.Linear(D,vocab_size,bias=False)
        self.apply(self._init_weights); self.lm_head.weight = self.token_embedding.weight
        self.architecture["parameter_count"] = sum(p.numel() for p in self.parameters())
    @staticmethod
    def _init_weights(module):
        if isinstance(module,(nn.Linear,nn.Embedding)): nn.init.normal_(module.weight,0.0,0.02)
        if isinstance(module,nn.Linear) and module.bias is not None: nn.init.zeros_(module.bias)
        if isinstance(module,nn.LayerNorm): nn.init.ones_(module.weight); nn.init.zeros_(module.bias)
    def forward(self, idx, targets=None, return_attentions=False):
        if idx.ndim != 2: raise ValueError("Token input must have rank 2: [B,T].")
        if idx.dtype not in (torch.int8,torch.int16,torch.int32,torch.int64,torch.uint8): raise TypeError("Token input must use an integer dtype.")
        if idx.numel() and (idx.min().item() < 0 or idx.max().item() >= self.vocab_size): raise ValueError(f"Token IDs must be in [0,{self.vocab_size-1}].")
        if idx.size(1) > self.architecture["block_size"]: raise ValueError(f"Context length {idx.size(1)} exceeds block_size={self.architecture['block_size']}; crop only in generation.")
        if targets is not None and (targets.shape != idx.shape or targets.dtype != torch.long): raise ValueError("Targets must be torch.long with the same [B,T] shape as input.")
        x = self.dropout(self.position(self.token_embedding(idx)))
        maps=[]
        for block in self.blocks:
            if return_attentions: x,w = block(x,True); maps.append(w)
            else: x = block(x)
        logits = self.lm_head(self.ln_f(x))
        loss = None if targets is None else F.cross_entropy(logits.reshape(-1,self.vocab_size),targets.reshape(-1))
        return {"logits": logits, "loss": loss, "attentions": maps if return_attentions else None}

# ---- preserved source boundary ----

seed_everything(CONFIG["seed"])
canonical_architecture_probe = FinanceMiniGPT(VOCAB_SIZE, dict(CONFIG["architecture"])).to(DEVICE)
canonical_architecture_probe.eval()
canonical_total = sum(parameter.numel() for parameter in canonical_architecture_probe.parameters())
assert canonical_total == canonical_architecture_probe.architecture["parameter_count"] == 3_178_240
assert canonical_architecture_probe.lm_head.weight.data_ptr() == canonical_architecture_probe.token_embedding.weight.data_ptr()

parameter_groups = {"token embedding / tied output": 0, "attention projections": 0, "feed-forward networks": 0, "normalization": 0, "remaining": 0}
for name, parameter in canonical_architecture_probe.named_parameters():
    if name == "token_embedding.weight": group = "token embedding / tied output"
    elif ".attn." in name: group = "attention projections"
    elif ".ffn." in name: group = "feed-forward networks"
    elif ".ln" in name or name.startswith("ln_f."): group = "normalization"
    else: group = "remaining"
    parameter_groups[group] += parameter.numel()
assert sum(parameter_groups.values()) == canonical_total

def _batch2_table(headers, rows):
    head = "| " + " | ".join(headers) + " |"
    rule = "| " + " | ".join("---" for _ in headers) + " |"
    body = ["| " + " | ".join(str(value) for value in row) + " |" for row in rows]
    return "\n".join([head, rule, *body])

from IPython.display import Markdown, display
architecture_rows = [
    ("Vocabulary size", VOCAB_SIZE), ("Block size", CONFIG["architecture"]["block_size"]),
    ("Embedding dimension", CONFIG["architecture"]["d_model"]), ("Attention heads", CONFIG["architecture"]["num_heads"]),
    ("Transformer blocks", CONFIG["architecture"]["num_layers"]),
    ("Feed-forward width / expansion", f'{CONFIG["architecture"]["d_ff"]} / {CONFIG["architecture"]["d_ff"] / CONFIG["architecture"]["d_model"]:.0f}×'),
    ("Dropout", CONFIG["architecture"]["dropout"]), ("Trainable parameters", f"{canonical_total:,}"),
    ("Weight tying", "token embedding and bias-free LM head share storage"),
    ("Initialization", "Normal(0, 0.02) weights; zero linear biases; unit LayerNorm scales"),
]
display(Markdown("### Canonical architecture\n\n" + _batch2_table(["Setting", "Value"], architecture_rows)))
parameter_rows = [(name, f"{count:,}") for name, count in parameter_groups.items()] + [("Total trainable parameters", f"{canonical_total:,}")]
display(Markdown("### Parameter accounting (tied weight counted once)\n\n" + _batch2_table(["Group", "Parameters"], parameter_rows)))

# ---- preserved source boundary ----

scientific_config_before = json.dumps(CONFIG, sort_keys=True)
corpus_identity_before = (manifest["corpus_id"], manifest["corpus_sha256"], len(documents), len(train_docs), len(val_docs))
canonical_paths = sorted((CORPUS_DIR / "results" / "canonical").iterdir())
canonical_hashes_before = {path.name: hashlib.sha256(path.read_bytes()).hexdigest() for path in canonical_paths if path.is_file()}

TINY_ARCH = {"block_size": 16,"d_model": 32,"num_heads": 4,"num_layers": 2,"d_ff": 64,"dropout": 0.0}
seed_everything(0); tiny_model = FinanceMiniGPT(VOCAB_SIZE,TINY_ARCH).to(DEVICE)
sample = torch.tensor([encode(train_docs[0]["body"][:12])],dtype=torch.long,device=DEVICE)
targets = torch.tensor([encode(train_docs[0]["body"][1:13])],dtype=torch.long,device=DEVICE)
assert sample.shape == targets.shape == (1, 12)
assert decode(sample[0].tolist())[1:] == decode(targets[0].tolist())[:-1]

result=tiny_model(sample,targets)
record_check("logits shape", result["logits"].shape==(1,12,VOCAB_SIZE))
record_check("finite scalar loss", result["loss"].ndim==0 and torch.isfinite(result["loss"]))
manual_shifted_loss = F.cross_entropy(result["logits"].reshape(-1,VOCAB_SIZE), targets.reshape(-1))
assert torch.allclose(result["loss"], manual_shifted_loss)
assert torch.isfinite(result["logits"]).all()

tiny_model.zero_grad(set_to_none=True); result["loss"].backward()
grads=[p.grad for p in tiny_model.parameters() if p.grad is not None]
record_check("finite nonzero gradients", grads and all(torch.isfinite(g).all() for g in grads) and any(torch.count_nonzero(g)>0 for g in grads))
optimizer=torch.optim.AdamW(tiny_model.parameters(),lr=1e-3); before=[p.detach().clone() for p in tiny_model.parameters()]
optimizer.step()
record_check("optimizer step changes parameters", any(not torch.equal(a,b) for a,b in zip(before,tiny_model.parameters())))
tiny_model.eval(); a=sample.clone(); b=sample.clone(); b[0,-1]=(b[0,-1]+1)%VOCAB_SIZE
with torch.no_grad(): la=tiny_model(a)["logits"]; lb=tiny_model(b)["logits"]
record_check("causal non-leakage", torch.equal(la[:,:-1],lb[:,:-1]))
overflow_raised = False
try: tiny_model(torch.zeros(1,17,dtype=torch.long,device=DEVICE))
except ValueError as exc: overflow_raised = "block_size" in str(exc)
record_check("context overflow rejection", overflow_raised)
record_check("weight tying storage", tiny_model.lm_head.weight.data_ptr()==tiny_model.token_embedding.weight.data_ptr())
record_check("serialized architecture parameter count", tiny_model.architecture["parameter_count"] == sum(p.numel() for p in tiny_model.parameters()))

seed_everything(7)
init_model = FinanceMiniGPT(VOCAB_SIZE,TINY_ARCH)
initialized_weights = [m.weight.detach().flatten() for m in init_model.modules() if isinstance(m, (nn.Linear, nn.Embedding))]
all_initialized = torch.cat(initialized_weights)
init_statistics_ok = abs(float(all_initialized.mean())) < 0.01 and 0.012 < float(all_initialized.std()) < 0.028
zero_linear_biases = all(torch.count_nonzero(m.bias) == 0 for m in init_model.modules() if isinstance(m, nn.Linear) and m.bias is not None)
record_check("GPT-style initialization statistics", init_statistics_ok, f"mean={float(all_initialized.mean()):.5f}, std={float(all_initialized.std()):.5f}")
record_check("zero linear biases", zero_linear_biases)

dropout_arch = {**TINY_ARCH, "dropout": 0.5}
seed_everything(11); dropout_model = FinanceMiniGPT(VOCAB_SIZE,dropout_arch).to(DEVICE)
dropout_model.train()
with torch.no_grad(): train_a=dropout_model(sample)["logits"]; train_b=dropout_model(sample)["logits"]
dropout_model.eval()
with torch.no_grad(): eval_a=dropout_model(sample)["logits"]; eval_b=dropout_model(sample)["logits"]
record_check("dropout train stochasticity", not torch.equal(train_a,train_b))
record_check("dropout eval determinism", torch.equal(eval_a,eval_b))

assert json.dumps(CONFIG, sort_keys=True) == scientific_config_before
assert (manifest["corpus_id"], manifest["corpus_sha256"], len(documents), len(train_docs), len(val_docs)) == corpus_identity_before
canonical_hashes_after = {path.name: hashlib.sha256(path.read_bytes()).hexdigest() for path in canonical_paths if path.is_file()}
assert canonical_hashes_after == canonical_hashes_before
print({"input_ids": tuple(sample.shape), "targets": tuple(targets.shape), "logits": tuple(result["logits"].shape), "loss": float(result["loss"].detach().cpu()), "canonical_parameters": canonical_total})
print("Architecture evidence PASS: shifted loss, finite logits/gradients, causality, tying, initialization, guards, parameter count, and scientific freeze")
```


## 9. Optimization and chronological validation

Training uses AdamW with explicit decay/no-decay parameter groups, a warmup-plus-cosine learning-rate schedule, and exhaustive chronological validation. Checkpoint selection is based on validation cross-entropy rather than training loss. Perplexity is reported only as a transformation of cross-entropy, not as an independent measure of economic meaning.

The visible checks below trace the AdamW parameter groups, learning-rate schedule, exhaustive validation, and guarded training step.


```python
def adamw_parameter_groups(model, weight_decay):
    decay=[]; no_decay=[]; membership={}
    for name,p in model.named_parameters():
        if not p.requires_grad: continue
        use_decay=p.ndim>=2 and name!="token_embedding.weight"
        (decay if use_decay else no_decay).append(p); membership[name]="decay" if use_decay else "no_decay"
    decay_ids,no_decay_ids={id(p) for p in decay},{id(p) for p in no_decay}; trainable_ids={id(p) for p in model.parameters() if p.requires_grad}
    if decay_ids & no_decay_ids: raise AssertionError("AdamW parameter groups overlap.")
    if decay_ids | no_decay_ids != trainable_ids: raise AssertionError("AdamW parameter groups do not cover every trainable parameter exactly once.")
    return [{"params":decay,"weight_decay":weight_decay,"group_name":"decay"},{"params":no_decay,"weight_decay":0.0,"group_name":"no_decay"}],membership

def configure_adamw(model,lr,weight_decay,betas=(0.9,0.95)):
    groups,membership=adamw_parameter_groups(model,weight_decay)
    return torch.optim.AdamW(groups,lr=lr,betas=betas),membership

def cosine_lr(step,total_steps,warmup_steps,max_lr,min_lr):
    """Warmup step 0 uses max_lr/warmup_steps; zero warmup starts at max_lr. The final optimization step is min_lr."""
    if total_steps<1: raise ValueError("total_steps must be positive.")
    if not 0<=step<total_steps: raise ValueError("step must be in [0, total_steps).")
    if not 0<=warmup_steps<total_steps: raise ValueError("warmup_steps must be in [0, total_steps).")
    if not (math.isfinite(max_lr) and math.isfinite(min_lr) and 0<=min_lr<=max_lr): raise ValueError("Learning rates must be finite with 0 <= min_lr <= max_lr.")
    if warmup_steps>0 and step<warmup_steps: return max_lr*(step+1)/warmup_steps
    if total_steps==1: return min_lr
    denominator=total_steps-1-warmup_steps
    if denominator<=0: return min_lr
    progress=(step-warmup_steps)/denominator
    return min_lr+0.5*(max_lr-min_lr)*(1+math.cos(math.pi*progress))

# ---- preserved source boundary ----

canonical_lr_steps = [0, CANONICAL_TRAINING["warmup_steps"] - 1, CANONICAL_TRAINING["warmup_steps"], 750, CANONICAL_TRAINING["steps"] - 1]
canonical_lr_rows = []
for step in canonical_lr_steps:
    lr_value = cosine_lr(step, CANONICAL_TRAINING["steps"], CANONICAL_TRAINING["warmup_steps"], CANONICAL_TRAINING["lr"], CANONICAL_TRAINING["min_lr"])
    canonical_lr_rows.append((step, f"{lr_value:.10f}"))
assert math.isclose(float(canonical_lr_rows[0][1]), CANONICAL_TRAINING["lr"] / CANONICAL_TRAINING["warmup_steps"], rel_tol=0, abs_tol=5e-11)
assert math.isclose(float(canonical_lr_rows[1][1]), CANONICAL_TRAINING["lr"], rel_tol=0, abs_tol=5e-11)
assert math.isclose(float(canonical_lr_rows[2][1]), CANONICAL_TRAINING["lr"], rel_tol=0, abs_tol=5e-11)
assert math.isclose(float(canonical_lr_rows[-1][1]), CANONICAL_TRAINING["min_lr"], rel_tol=0, abs_tol=5e-11)
display(Markdown("**Canonical schedule boundary evidence**\n\n" + markdown_table(["Optimization step", "Learning rate"], canonical_lr_rows)))

# ---- preserved source boundary ----

def safe_perplexity(cross_entropy):
    if not math.isfinite(cross_entropy): raise FloatingPointError("Cross-entropy must be finite before perplexity.")
    return math.exp(cross_entropy) if cross_entropy < math.log(np.finfo(np.float64).max) else float("inf")

@torch.no_grad()
def evaluate_exhaustive_details(model,dataset,batch_size,device):
    prior=model.training; model.eval(); weighted=0.0; tokens=0
    try:
        for start in range(0,len(dataset),batch_size):
            pairs=[dataset[i] for i in range(start,min(start+batch_size,len(dataset)))]
            x=torch.stack([p[0] for p in pairs]).to(device); y=torch.stack([p[1] for p in pairs]).to(device)
            loss=model(x,y)["loss"]
            if not torch.isfinite(loss): raise FloatingPointError("Non-finite exhaustive validation loss.")
            weighted+=float(loss.detach().cpu())*y.numel(); tokens+=y.numel()
        ce=weighted/tokens
        return {"cross_entropy":ce,"perplexity":safe_perplexity(ce),"windows":len(dataset),"tokens":tokens}
    finally: model.train(prior)

def evaluate_exhaustive(model,dataset,batch_size,device): return evaluate_exhaustive_details(model,dataset,batch_size,device)["cross_entropy"]

def enter_training_mode(model,allowed):
    if not allowed: raise RuntimeError("Training requires an authorized resolved run configuration.")
    model.train()

def train_model(model,train_ds,val_ds,training,checkpoint_context,device):
    enter_training_mode(model,training.get("authorized_training") is True and training.get("run_mode") in {"phase3b-smoke","phase4-canonical"})
    opt,_=configure_adamw(model,training["lr"],training["weight_decay"],tuple(training["betas"])); gen=torch.Generator().manual_seed(training["seed"])
    metrics=[]; best_loss=float("inf"); best_step=None; tokens=0; started=time.perf_counter(); validation_steps=set(training["validation_steps"])
    for step in range(training["steps"]):
        lr=cosine_lr(step,training["steps"],training["warmup_steps"],training["lr"],training["min_lr"])
        for group in opt.param_groups: group["lr"]=lr
        x,y=train_ds.sample_batch(training["batch_size"],gen,device); opt.zero_grad(set_to_none=True)
        loss=model(x,y)["loss"]
        if not torch.isfinite(loss): raise FloatingPointError(f"Non-finite training minibatch loss at step {step}")
        loss.backward(); grads=[p.grad for p in model.parameters() if p.grad is not None]
        if not grads or not all(torch.isfinite(g).all() for g in grads): raise FloatingPointError(f"Non-finite/missing gradients at step {step}")
        grad_norm=torch.nn.utils.clip_grad_norm_(model.parameters(),training["grad_clip"])
        if not torch.isfinite(grad_norm): raise FloatingPointError(f"Non-finite gradient norm at step {step}")
        opt.step(); tokens+=y.numel()
        elapsed=time.perf_counter()-started
        row={"step":step,"train_minibatch_ce":float(loss.detach().cpu()),"gradient_norm_before_clipping":float(grad_norm.detach().cpu()),"validation_ce":None,"validation_perplexity":None,"lr":lr,"tokens_processed":tokens,"elapsed_seconds":elapsed,"throughput_tokens_per_second":tokens/elapsed if elapsed>0 else None}
        if step in validation_steps:
            validation=evaluate_exhaustive_details(model,val_ds,training["eval_batch_size"],device)
            row.update({"validation_ce":validation["cross_entropy"],"validation_perplexity":validation["perplexity"]})
            metrics.append(row)
            if validation["cross_entropy"]<best_loss:
                best_loss,best_step=validation["cross_entropy"],step
                save_checkpoint(training["best_path"],"best",model,opt,None,training,checkpoint_context,step,best_step,best_loss,metrics)
            continue
        metrics.append(row)
    if best_step is None: raise RuntimeError("No validation checkpoint was produced.")
    save_checkpoint(training["final_path"],"final",model,opt,None,training,checkpoint_context,training["steps"]-1,best_step,best_loss,metrics)
    return {"metrics":metrics,"best_step":best_step,"best_validation_loss":best_loss,"tokens_processed":tokens,"elapsed_seconds":time.perf_counter()-started}
```

## 13. Controlled autoregressive decoding

Generation first reconstructs the best checkpoint, encodes a fixed prompt, and repeatedly keeps only the most recent 128 characters. It obtains next-character logits, chooses or samples one character, appends it, and repeats. Greedy decoding (`temperature=0`) selects the maximum-logit character deterministically. Stochastic decoding divides logits by temperature 0.8, retains only the 20 largest candidates, renormalizes with softmax, and samples with a fixed seed. Higher temperature flattens the candidate distribution; lower positive temperature sharpens it. Top-k limits the candidate set but does not certify quality or factuality.

The table records every canonical prompt and decoding configuration. The 200-character samples shown afterward are loaded unchanged from the canonical JSON; the short executable probes check decoding mechanics without replacing those artifacts. Seeded stochastic equality is expected on the same PyTorch/device backend, while exact sampling can differ across backends.

## 16. Findings, limitations, and conclusion

### Direct answer to the research question

The compact model successfully engineers an end-to-end Transformer pipeline and learns measurable next-character regularities plus aspects of FOMC surface form. The recorded loss and perplexity establish predictive improvement under the character tokenizer; the frozen generations establish that those gains do **not** reliably yield sustained semantic monetary-policy coherence. This limitation is consistent with the joint constraints of corpus size, tokenization, context, capacity, and training budget. The result demonstrates the Transformer pipeline honestly rather than proving a production language model.

### What was and was not established

| Evidence layer | Supported conclusion | Unsupported inference |
|---|---|---|
| Engineering | Causal attention, training, exhaustive validation, checkpointing, strict reload, and artifact reconciliation work as specified. | That implementation correctness guarantees model quality. |
| Predictive | Validation CE reached 1.891652 and character perplexity 6.630312 at the selected checkpoint. | Semantic understanding, forecasting skill, or production readiness. |
| Generated text | Some local character patterns and FOMC-like surface fragments appear; degeneration and malformed text remain visible. | Reliable policy stance, factuality, economic reasoning, or coherent long-form language. |

### Limitations

1. The training set contains only 82 statements and 237,840 characters.
2. Adjacent overlapping windows are strongly dependent and are not independent documents or economic observations.
3. Character tokenization must learn spelling, word boundaries, syntax, and semantics from individual characters.
4. Context length 128 captures mainly local text, not extended policy discourse.
5. The model is compact (3,178,240 parameters) and the compute budget is limited.
6. The 2025 validation split was reused for checkpoint selection and analysis.
7. There is no pristine final test set.
8. Nothing here supports production forecasting, policy inference, trading use, or a claim of economic reasoning.

### Most informative next experiments

New, separately governed studies could extend statement-only history, broaden the corpus to other Federal Reserve communications, adopt subword tokenization, lengthen context, scale capacity appropriately, and run controlled ablations that hold other variables as comparable as possible. Those experiments should introduce a fresh final test protocol rather than repeatedly tuning against 2025. They are not performed here, and this capstone does not silently change the frozen corpus or model.
